# Trabajo Práctico: Sistema de Localización y Segmentación de Ojos en Tiempo Real

## Objetivo

En este trabajo deberán desarrollar un sistema de visión artificial en tiempo real utilizando la cámara web de una computadora. El sistema deberá combinar tres paradigmas fundamentales vistos durante el curso:

* Clasificación de imágenes
* Detección de objetos
* Segmentación de instancias

Cada etapa utilizará la salida de la anterior para reducir la cantidad de información que debe procesarse y mejorar la eficiencia del sistema.

El sistema deberá seguir el siguiente flujo de procesamiento:

1. Tomar imágenes de manera continua con su cámara web.
2. Aplicar un modelo de clasificación para la detección o no de un rostro en pantalla.
3. Solo en caso de que exista un rostro en pantalla, deberá activar un módulo que incluya un modelo de detección que indique dónde se encuentra el rostro.
4. Habiendo localizado el bounding box que enmarca la cara, deberá realizar un recorte sobre esta área de interés y pasarla al siguiente módulo.
5. Sobre el recorte de la imagen, aplicar un modelo de segmentación para localizar cada ojo por separado.
6. Luego de obtenida la máscara, deberá dibujarla sobre la imagen original.

---

## Etapa 1 – Clasificación
En esta etapa se debe determinar si la imagen contiene un rostro o no. La salida deberá ser una de las siguientes clases:

* Rostro presente
* Sin rostro

Si la respuesta es Sin rostro, el sistema no deberá ejecutar las siguientes etapas.

### Requisitos

La clasificación deberá ejecutarse sobre el cuadro completo proveniente de la cámara.

Se deberá mostrar en pantalla un texto indicando:

- Si hay un rostro presente
- Confianza

### Modelos sugeridos

Se recomienda utilizar alguno de los siguientes modelos preentrenados:

> MobileNetV3 Small (recomendado)

- Extremadamente rápido
- Ideal para tiempo real
- Pocas operaciones

> EfficientNet-B0

- Mayor precisión

> ResNet18

- Muy sencilla de utilizar y ampliamente documentada

> ConvNeXt Tiny

- Solo recomendado para equipos con GPU. Alta velocidad

## Etapa 2 – Detección

Tras determinar la presencia de un rostro deberá localizar su ubicación y enmarcarlo en una bounding box.

Sobre la imagen deberá dibujar:

- Bounding Box
- Confianza

Posteriormente deberá recortarse el ROI correspondiente al rostro.

### Modelos sugeridos

> YOLOv8n-face (recomendado)

- Muy rápido.

> RetinaFace

 -Excelente precisión.

> BlazeFace

- Muy eficiente.

> SSD Face Detector

- Alternativa sencilla.

## Etapa 3 – Segmentación de instancias

Sobre el ROI obtenido anteriormente, segmentar de manera independiente:

- El ojo izquierdo
- El ojo derecho

No basta con segmentar una única región denominada "ojos"; el sistema deberá diferenciar ambas instancias. Tras obtener cada máscara, dibújela en la imagen original.

### Modelos sugeridos

> YOLOv8-Seg ⭐

- Muy sencillo de utilizar.

> Mask R-CNN

- Excelente opción para comprender segmentación de instancias.

> SAM 2

- Puede utilizarse únicamente si se automatiza correctamente la selección del ROI.
